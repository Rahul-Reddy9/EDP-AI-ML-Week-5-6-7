Train an XGBoost classifier on the processed split, using `scale_pos_weight` to compensate for the heavy class imbalance.

```
import sys
sys.path.append('..')
import os
os.chdir('..')
from src.train_xgboost import load_processed_data, train_xgboost
import joblib
```

```
X_train, X_test, y_train, y_test = load_processed_data()
X_train.shape, X_test.shape
```
