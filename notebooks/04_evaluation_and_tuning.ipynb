# 04 - Evaluation & Tuning (Weeks 6-7)
Evaluate Random Forest and XGBoost (accuracy, precision, recall, F1, confusion matrix), tune both with `GridSearchCV`, and compare all four models.

```
import sys
sys.path.append('..')
import os
os.chdir('..')
import joblib
from src.evaluate import load_processed_data, evaluate_model
from src.tune import tune_random_forest, tune_xgboost
```

```
X_train, X_test, y_train, y_test = load_processed_data()
```

## Evaluate the untuned models

```
rf = joblib.load('models/random_forest.pkl')
xgb = joblib.load('models/xgboost.pkl')
rf_metrics = evaluate_model(rf, 'random_forest', X_test, y_test)
xgb_metrics = evaluate_model(xgb, 'xgboost', X_test, y_test)
rf_metrics, xgb_metrics
```

## Hyperparameter tuning

```
best_rf = tune_random_forest(X_train, y_train)
joblib.dump(best_rf, 'models/tuned_random_forest.pkl')
```

```
best_xgb = tune_xgboost(X_train, y_train)
joblib.dump(best_xgb, 'models/tuned_xgboost.pkl')
```

## Evaluate tuned models & compare

```
tuned_rf_metrics = evaluate_model(best_rf, 'tuned_random_forest', X_test, y_test)
tuned_xgb_metrics = evaluate_model(best_xgb, 'tuned_xgboost', X_test, y_test)
```

```
import pandas as pd
rows = []
for name, m in [('Random Forest', rf_metrics), ('Tuned Random Forest', tuned_rf_metrics), ('XGBoost', xgb_metrics), ('Tuned XGBoost', tuned_xgb_metrics)]:
    row = {'Model': name}
    row.update({k.replace('_',' ').title(): round(v,4) for k, v in m.items()})
    rows.append(row)
comparison_df = pd.DataFrame(rows)
comparison_df.to_csv('results/comparison.csv', index=False)
comparison_df
```
